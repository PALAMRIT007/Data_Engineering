#### get row level duplicates 

with t1 as (
select *, row_number() over(partition by col1,col2,col3...coln) as rnk )
from t2
)

select * from t1 where rnk>1;

###### we can skip using order by, and need to mention all the columns present in table if want to get row level duplicates

#### get the records from ta which are not prsent in tb

select ta.*
from ta a
left join tb b
on a.id = b.id
and a.col2 - b.col2
where b.id is null
